```python
import pandas as pd
import numpy as np
```

### GDP CLEANER

```python
gdppc = pd.read_csv("matched_gdp.csv")
gdppc.replace("N/A", np.nan, inplace=True)
gdppc = gdppc.drop(columns=["Indicator Name", "Indicator Code"])
gdppc_long = gdppc.melt(id_vars=["Country Code"], var_name="Year", value_name="gdp_per_capita")
gdppc_long['Year'] = gdppc_long['Year'].astype(str)
gdppc_long_cleaned = gdppc_long[~gdppc_long['Year'].str.contains("Unnamed", na=False)]
gdppc_long_cleaned['Year'] = pd.to_numeric(gdppc_long_cleaned['Year'], errors='coerce', downcast='integer')
gdppc_long_cleaned = gdppc_long_cleaned.dropna(subset=['Year'])
gdppc_long_cleaned['Year'] = gdppc_long_cleaned['Year'].astype(int)
gdppc_long_cleaned = gdppc_long_cleaned.dropna(subset=['Country Code'])
gdppc_long_cleaned.to_csv("wb_gdp_per_capita.csv", index=False)
gdppc_long_cleaned
```
